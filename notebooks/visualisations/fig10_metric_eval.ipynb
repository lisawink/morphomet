```python
# Extract columns and rows
columns = data[0]
rows = data[1:]

# Build DataFrame
df = pd.DataFrame(
    [values for _, values in rows],
    index=[name for name, _ in rows],
    columns=columns
)

df
```

```python
# cumulative bar plot with rows as x - axis and columns as bars
# x axis labels horizontal
df.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.xlabel('Urban Morphological Metric')
plt.ylabel('Value')
plt.title('Cumulative Bar Plot of Metrics')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.xticks(rotation=0)
plt.tight_layout()
```

```python
df_quant = df[[df.columns[0], df.columns[1], df.columns[2], df.columns[6], df.columns[7]]]
# reorder df by overall score big to small
df_quant['Overall Score'] = df_quant[df_quant.columns[0]]+df_quant[df_quant.columns[1]]+df_quant[df_quant.columns[2]]+df_quant[df_quant.columns[3]]+df_quant[df_quant.columns[4]]
df_quant = df_quant.sort_values(by='Overall Score', ascending=False)
df_quant.drop(columns=['Overall Score'], inplace=True)  # Remove the overall score column if not needed for plotting
```

```python
order = df_quant.std().sort_values().index  # low variance → bottom
bottom = np.zeros(len(df_quant))
x = np.arange(len(df_quant))

for col in order:
    plt.bar(x, df_quant[col], bottom=bottom, label=col)
    bottom += df_quant[col].values
plt.xticks(x, df_quant.index, rotation=90)
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles[::-1], labels[::-1], loc='upper left', bbox_to_anchor=(1, 1))
plt.ylabel('Cumulative Score')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator, FuncFormatter
import numpy as np

# --- custom formatter to remove decimals like 0.0 ---
def int_formatter(x, pos):
    return f"{int(x)}" if x.is_integer() else f"{x:.0f}"

# make (a) and (b) smaller than (c)
fig, axs = plt.subplots(
    nrows=3, ncols=1,
    figsize=(7, 12),
    gridspec_kw={'height_ratios': [1, 1, 2]}
)

# ---------- (a) Quantitative ----------
order = df_quant.std().sort_values().index
bottom = np.zeros(len(df_quant))
x = np.arange(len(df_quant))

for col in order:
    axs[0].bar(x, df_quant[col], bottom=bottom, label=col)
    bottom += df_quant[col].values

axs[0].set_xticks(x)
axs[0].set_xticklabels(df_quant.index, rotation=45, fontsize=12)
axs[0].set_ylabel('Cumulative Score\n(Quantitative)', fontsize=14)

axs[0].yaxis.set_major_locator(MaxNLocator(integer=True))
axs[0].yaxis.set_major_formatter(FuncFormatter(int_formatter))
axs[0].grid()

# ---------- (b) Qualitative ----------
order2 = ['Data\navailability', 'Computational\nspeed', 'Robustness', 'Physical\ninterpretability']
bottom = np.zeros(len(df_sub))
x = np.arange(len(df_sub))
colors = plt.cm.tab10.colors[5:9]

for col, color in zip(order2, colors):
    axs[1].bar(x, df_sub[col], bottom=bottom, label=col, color=color)
    bottom += df_sub[col].values

axs[1].set_xticks(x)
axs[1].set_xticklabels(df_sub.index, rotation=45, fontsize=12)
axs[1].set_ylabel('Cumulative Score\n(Qualitative)', fontsize=14)

axs[1].yaxis.set_major_locator(MaxNLocator(integer=True))
axs[1].yaxis.set_major_formatter(FuncFormatter(int_formatter))
axs[1].set_ylim(0, 5)  # Set y-axis limit to 4 for better visualization
axs[1].grid()

# ---------- (c) All ----------
order_all = order.to_list() + order2
bottom = np.zeros(len(df))
x = np.arange(len(df))

for col in order_all:
    axs[2].bar(x, df[col], bottom=bottom, label=col)
    bottom += df[col].values

axs[2].set_xticks(x)
axs[2].set_xticklabels(df.index, rotation=45, fontsize=12)
axs[2].set_ylabel('Cumulative Score (All)', fontsize=14)

axs[2].yaxis.set_major_locator(MaxNLocator(integer=True))
axs[2].yaxis.set_major_formatter(FuncFormatter(int_formatter))
axs[2].grid()

# --- get handles from first (quantitative) plot ---
h1, l1 = axs[0].get_legend_handles_labels()

# --- get handles from second (qualitative) plot ---
h2, l2 = axs[1].get_legend_handles_labels()

# combine in desired order: first plot, then second plot
handles = h2 + h1
labels = l2 + l1
fig.legend(
    handles[::-1], labels[::-1],
    loc='upper center',
    bbox_to_anchor=(0.55, 1.03),
    ncol=3,
    fontsize=12,
)

for ax in axs:
    ax.tick_params(axis='y', labelsize=12)
    #ax.set_xlim(-1.8, len(x) - 0.2)
"""
# ---- custom key (symbols + definitions) ----
key_text = (
    r'$\it{\lambda_P}$ - Plan Area Fraction | $\it{\lambda_V}$ - Volume Density | $\it{\lambda_E}$ - Facet Density | $\it{H}$ - Height' '\n'
    r'$\it{W}$ - Width | $\it{SVF}$ - Sky View Factor |  $\it{H:W_1}$–$\it{H:W_3}$ - Height–Width ratios' '\n'
    r'$\it{SC}$ - Shape Complexity | $\it{Adj}$ - Adjacency | $\it{C}$ - Connectivity'
)

fig.text(
    0.5, -0.01, key_text,
    ha='center', va='top',
    fontsize=11,
    wrap=True
)
"""

from matplotlib.lines import Line2D

# ---- clean legend-style key ----
key_labels = [
    r'$\it{\lambda_P}$ - Plan Area Fraction (m$^2$ m$^{-2}$)',
    r'$\it{\lambda_V}$ - Building Volume Density (m$^3$ m$^{-2}$)',
    r'$\it{\lambda_E}$ - Exposed Facet Density (m$^2$ m$^{-2}$)',
    r'$\it{H}$ - Height (m)',
    r'$\it{W}$ - Width (m)',
    r'$\it{SVF}$ - Sky View Factor',
    r'$\it{H:W_1}$–$\it{H:W_3}$ - Height:Width ratios',
    r'$\it{SC}$ - Shape Complexity',
    r'$\it{Adj}$ - Adjacency',
    r'$\it{C}$ - Connectivity'
]

# dummy handles (no markers)
handles = [Line2D([0], [0], color='none') for _ in key_labels]

fig.legend(
    handles, key_labels,
    loc='lower center',
    bbox_to_anchor=(0.52, -0.04),
    ncol=2,
    fontsize=12,
    handlelength=0,
    handletextpad=0
)

from matplotlib.patches import Patch

legend_elements = [
    Patch(facecolor='olive', label='Physical interpretability: indirect (0), direct (1)'),

    Patch(facecolor='gray', label='Robustness: low (0.2), medium (0.6), high (1)'),

    Patch(facecolor='tab:pink', label='Computational speed: low (0.2), medium (0.6), high (1)'),

    Patch(facecolor='sienna', label='Data availability: 3D (0.2), 2.5D (0.6), 2D (1)')
]

fig.legend(
    handles=legend_elements,
    loc='upper center',
    bbox_to_anchor=(0.5, 0.7),
    ncol=1,
    frameon=False,
    fontsize=11
)



plt.tight_layout(rect=[0, 0.08, 1, 0.92])
plt.subplots_adjust(hspace=0.30)
plt.savefig('../../figures/fig10/cumulative_bar_plot.png', dpi=300, bbox_inches='tight')
plt.savefig('../../figures/fig10/cumulative_bar_plot.pdf', bbox_inches='tight')
plt.savefig('../../figures/fig10/cumulative_bar_plot.svg', bbox_inches='tight')
```

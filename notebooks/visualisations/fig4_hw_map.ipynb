```python
# HW ratio map

radii = [30,40,50,60,70,80,90,100,110,120,130,140,150,160,170,180,190,200,210,220,230,240,250,260,270,280,290,300,320,340,360,380,400,450,500]

stats = {}

for radius in radii:
    print(radius)
    gdf = gpd.read_parquet(f'../../data/processed_data/processed_station_params_{radius}.parquet')
    stats[radius] = gdf['StrHW_median'].max()
    gdf.index = gdf['station_id']
    stats[radius] = gdf['StrHW_wmean']
```

```python
stations = gdf['station_id'].unique()
```

```python
stats
```

```python
for station in stations:
    plt.plot(radii, [stats[radius][station] for radius in radii], label=station)
    plt.legend()
```

```python
range_list = []

for station in stations:
    print(station)
    # difference between max and min value of the hw ratio across all radii
    range = np.nanmax([stats[radius][station] for radius in radii]) - np.nanmin([stats[radius][station] for radius in radii])
    range_list.append(range)
    print(range)
```

```python
# check how many values in range list are above 0.1
print(sum([1 for range in range_list if range < 0.1]))
```
